```python
import pandas as pd
from io import StringIO
from datetime import datetime, timezone
import time
import random
```

```python
holiday_movies = pd.read_csv(
    "https://bcdanl.github.io/data/holiday_movies.csv"
)
display(holiday_movies)
```

```python
holiday_movies.info
```

```python
holiday_movie_genres = pd.read_csv(
    "https://bcdanl.github.io/data/holiday_movie_genres.csv"
)
display(holiday_movie_genres)
```

```python
holiday_movie_genres.info
```

This dataset contains 2265 movies that are holiday-themed.

## Sorting

```python
top_rated = (
    holiday_movies
    .sort_values("average_rating", ascending=False)
    .head(10)
)

top_rated[["primary_title", "year", "average_rating", "num_votes"]]
```

By sorting the data, the movie that has the highest average rating is "NLO Spirit of Christmas" with 10.0. Interestingly, the top ten highest average rating movies all contain a word "Christmas".

##Filtering

```python
recent_movies = holiday_movies[holiday_movies["year"] >= 2000]
recent_movies.shape
```

More than half of movies in the data set are released after year 2000.

##Joining & Counting

```python
holiday_full = pd.merge(
    holiday_movies,
    holiday_movie_genres,
    on="tconst",
    how="left"
)
holiday_full
```

```python
genre_counts = (
    holiday_full["genres"]
    .value_counts()
)

genre_counts
```

Comedy, drama, romance and Family dominate the genre distribution. This might be suggesting that holiday-themed movies primarily emphasize humor, emotion, and family-oriented storytelling rather than action or news.

##Indexing

```python
classic_movies = holiday_movies.loc[
    holiday_movies["year"] < 2000,
    ["primary_title", "year", "average_rating"]
]

classic_movies.sort_values("average_rating", ascending=False)
```

A movie that had the highest average rating by the year 2000 was Marie Osmond's Merry Christmas on 1986.

```python
classic_movies.sort_values("year", ascending=False)
```

The oldest holiday themed movie was released in 1929 called Sailor's Holiday.
